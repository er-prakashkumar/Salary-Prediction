# **3. Data Cleaning**

**Detect missing values**

```python
df.isnull().sum()
```

```python
# Droping the Salary Row becuase it is a target.
df = df.dropna(subset=["Salary"])
df.isnull().sum()
```

*Mean is pulled toward outliers, median is not — so for skewed/outlier-prone numeric data, impute with median*

```python
median_age = df["Age"].median()
mode_prating = df["PerformanceRating"].mode()[0]
df["Age"] = df["Age"].fillna(median_age)
df["PerformanceRating"] = df["PerformanceRating"].fillna(mode_prating)
```

*For ordinal/discrete scales, use mode, not median or mean — because you must fill in a value that genuinely exists on that scale.*

```python
df.isnull().sum()
```

```python
# Check duplicates while IGNORING the EmployeeID column
duplicate_check_cols = df.columns.drop('EmployeeID')

print("Number of duplicates (ignoring EmployeeID):",
      df.duplicated(subset=duplicate_check_cols).sum())

print(df[df.duplicated(subset=duplicate_check_cols, keep="first")])
# To show all duplicates (False)
# "first" show only duplicates ignore first
# "last"  show only duplicates ignore last

# Now actually remove them
df = df.drop_duplicates(subset=duplicate_check_cols)
print("Shape after removing true duplicates:", df.shape)
```

```python
# Step 1: Standardize casing - make everything Title Case
df['Education'] = df['Education'].str.strip().str.title()
# Step 2: Check what we have now
df['Education'].unique()
```

```python
# Step 3: Fix the spelling ('Master' -> 'Masters')
df['Education'] = df['Education'].replace('Master', 'Masters')

# Step 4: Final check
print(df['Education'].unique())
print(df['Education'].value_counts())
# bhej
# [REDACTED_TOKEN]
```

```python
# Step 1: Define a sane realistic boundary
df.loc[(df['Age'] < 18) | (df['Age'] > 65), 'Age'] = np.nan
```

```python
# Step 2: Recompute median AFTER removing outliers, then fill it in
median_age_clean = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age_clean)
```
